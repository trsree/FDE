```python
from langchain_groq import ChatGroq
llm = ChatGroq(api_key=groq_api_key, model=LLAMA_MODEL)
def simple_chat_without_memory(user_query):
    response = llm.invoke(user_query)
    return response.content
```

<h4>LLM calls with conversation history</h4>


```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
conversation = []
def simple_chat_with_memory(user_query):
    conversation.append(HumanMessage(content=user_query))
    response = llm.invoke(conversation)
    conversation.append(AIMessage(content=response.content))
    return response.content
```

```python
response1 = simple_chat_with_memory("I want to learn about AI")
print(response1)
```

```python
response2 = simple_chat_with_memory("Can you give me a learning plan?")
print(response2)
```

<h5> Chaining with LangChain</h5>

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate(messages=[
    ("system", "You are a helpful personal assistant."),
    ("human", "{user_query}"),
  
])
chain = prompt | llm
response_chain = chain.invoke({"user_query": "I decided to learn AI"})
print(response_chain.content)
```

```python
response_chain = chain.invoke({"user_query": "Can you give me a learning plan?"})
print(response_chain.content)
```

The above chain invoke doesnt keep memory. To maintain memory management with user sessions, Langchain provides us RunnableWithMessageHistory and ChatMessageHistory objects. ChatMessageHistory needs langchain-community package to be installed.

<h4>LangChain Session and Memory Management</h4>

```python
prompt_with_memory = ChatPromptTemplate(messages=[
    ("system", "You are a helpful personal assistant."),
    ("placeholder", "{chat_history}"),
    ("human", "{user_query}")
])
chain_with_memory = prompt_with_memory | llm
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory

store = {}
def get_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]
```

```python
# User Session with chat history
from langchain_core.runnables import RunnableWithMessageHistory
chat_with_memory = RunnableWithMessageHistory(
        runnable=chain_with_memory, 
        get_session_history=get_history,
        input_messages_key="user_query",
        history_messages_key="chat_history")
```

```python
response1 = chat_with_memory.invoke({"user_query": "I want to learn about AI"}, {"configurable": {"session_id": "user1" }})
print(response1.content)
```

```python
response2 = chat_with_memory.invoke({"user_query": "Can you give me a learning plan?"}, {"configurable": {"session_id": "user1" }})
print(response2.content)
```

```python
response1_user2 = chat_with_memory.invoke({"user_query": "Give me a learning plan?"}, {"configurable": {"session_id": "user2" }})
print(response1_user2.content)
```
