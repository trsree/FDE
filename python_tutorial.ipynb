<h4> Calculate the area and perimeter of rectangle </h4>

```python
def calculate_area_and_perimeter(len, width):
    """
    Args:
     len(int32)
     width(int32)
    Returns:
     str
    """
    # Write your code here.
    area = len * width
    per = 2*(len + width)
    return f'Area and Perimeter of rectangle is {area}, {per} respectively.'
    

calculate_area_and_perimeter(10, 5)
```

<h4>Find the maximum of three numbers </h4>

```python

def find_maximum_of_three(num1, num2, num3):
    """
    Args:
     num1(int32)
     num2(int32)
     num3(int32)
    Returns:
     int32
    """
    # Write your code here.
    if (num1 > num2) and (num1 > num3):
        max = 'num1'
    elif (num2 > num1) and (num2 > num3):
        max = 'num2'
    else:
        max = 'num3'
    return max
    
input = {
"num1": -34,
"num2": -200,
"num3": 0
}
print(find_maximum_of_three(input['num1'], input['num2'], input['num3']))
```

<h4> Find the sum of first n natural numbers </h4>

```python
def sum_of_natural_numbers(n):
    """
    Args:
     n(int32)
    Returns:
     int32
    """
    # Write your code here.
    sum = 0
    while (n >= 1):
        sum = n + sum
        n -= 1
    return sum
    
sum_of_natural_numbers(10)
```

<h4> Find if the given year is leap year or not </h4>

```python
def is_leap_year(year):
    """
    Args:
     year(int32)
    Returns:
     bool
     If the year is divisible by 4, it is a leap year.
    However, if that year is divisible by 100, it is not a leap year, unless...
    The year is also divisible by 400, in which case it is a leap year.
    """
    # Write your code here.
    if (year %  4 == 0) and (year % 100 != 0) :
        return True
    elif (year % 400 == 0):
        return True
    else:
        return False

print(is_leap_year(2024))
print(is_leap_year(2000))
print(is_leap_year(1900))
print(is_leap_year(2004))
print(is_leap_year(4000))
```

```python
def length_of_last_word(sentence):
    """
    Args:
     sentence(str)
    Returns:
     int32
    """
    # Write your code here.
    if sentence:
        sentence_lastword = sentence.split()[-1] if sentence.split() else ''
        return len(sentence_lastword)
    else:
        return 0
print(length_of_last_word(" "))
print(length_of_last_word("Hello World"))
print(length_of_last_word("The quick brown fox jumps over the lazy dog"))
```

<h4> Truncate the sentence </h4>

```python
def truncate_sentence(s, k):
    """
    Args:
     s(str)
     k(int32)
    Returns:
     str
    """
    # Write your code here.
    if len(s) > 0 and k > 0:
        words = s.split()
        truncated = words[:k]
        return ' '.join(truncated)
    return ''
print(truncate_sentence("Hello world, this is a test.", 4))
print(truncate_sentence("the quick brown fox", 5))
```

<h4> Count the number of substrings in a string </h4>

```python
def count_substrings(s, substr):
    """
    Args:
     s(str)
     substr(str)
    Returns:
     int32
     Complete the program that takes a string s and a substring substr as input and determines how many times the given substring substr appears within the string.

The program should account for overlapping substrings, meaning if a substring appears more than once starting from different positions in the string, each occurrence should be counted separately.
    """
    # Write your code here.
    count = 0
    start = 0
    while True:
        pos = s.find(substr, start)
        if pos == -1:
            break
        count += 1
        start = pos + 1
    return count
print(count_substrings("my name is name", "name"))
print(count_substrings("abcbcabcdabc", "abc"))
print(count_substrings("aaaaa", "aa"))
```

<h4> Find the left pivot index of the numbers from the list </h4>

```python
def get_pivot_index(numbers):
    """
    Args:
     numbers(list_int32)
    Returns:
     int32
     Given a list of integers numbers, calculate the pivot index of this list.

The pivot index is the index where the sum of all the numbers strictly to the left of the index is equal to the sum of all the numbers strictly to the index's right. If the index is on the left edge of the array, then the left sum is 0 because there are no elements to the left. This also applies to the right edge of the array.

Return the leftmost pivot index. If no such index exists, return -1.
    """
    # Write your code here.
    if len(numbers) == 0:
        return -1
    total_sum = sum(numbers)
    left_sum = 0

    for i, num in enumerate(numbers):
        right_sum = total_sum - left_sum - num
        if left_sum == right_sum:
            return i
        left_sum += num
    return -1

print(get_pivot_index([2, 3, 1, -1, 1, 1, 4]))
print(get_pivot_index([2, 3, 5]))

    
```

<h4>Grouping & Aggregation</h4>

```python
df = pd.DataFrame({
    'dept': ['Sales', 'Sales', 'IT', 'IT', 'HR'],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'salary': [50000, 60000, 70000, 65000, 55000]
})

# Group by and aggregate
grouped = df.groupby('dept')['salary'].mean()
# Sales    55000
# IT       67500
# HR       55000

# Multiple aggregations
agg_result = df.groupby('dept').agg({
    'salary': ['mean', 'min', 'max', 'count']
})

# Custom aggregation
custom = df.groupby('dept').apply(lambda x: x['salary'].max() - x['salary'].min())

# Pivot tables
pivot = df.pivot_table(
    values='salary',
    index='dept',
    aggfunc='mean'
)
```
